#Travel Demand & Destination Recommendation Prediction System

#Import libraries

```python
import pandas as pd
import numpy as np
```

#Load dataset

```python
df = pd.read_csv("../data/raw/travel_tourism_dataset.csv")

print("Original Shape:", df.shape)
```

1. Remove duplicate rows

```python
print("Duplicate rows before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after:", df.duplicated().sum())
```

2. Clean column names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("/", "_", regex=False)
)

print("\nCleaned Columns:")
print(df.columns.tolist())
```

3. Check missing values

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

4. Handle missing values

```python
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = df.select_dtypes(
    include=["object"]
).columns


# Numerical missing values → median
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())


# Categorical missing values → mode
for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])
```

5. Check data types

```python
print("\nData Types:")
print(df.dtypes)
```

6. Identify categorical columns

```python
print("\nCategorical Columns:")
print(df.select_dtypes(include="object").columns.tolist())
```

```python
df = pd.get_dummies(
    df,
    columns=df.select_dtypes(include="object").columns,
    drop_first=True
)
```

7. Identify numerical columns

```python

print("\nNumerical Columns:")
print(
    df.select_dtypes(
        include=["int64", "float64"]
    ).columns.tolist()
)
```

9. Final missing value check

```python
print("\nTotal Missing Values:",
      df.isnull().sum().sum())
```

10. Final duplicate check

```python
print("Total Duplicate Rows:",
      df.duplicated().sum())
```

 11. Save preprocessed dataset

```python
df.to_csv(
    "travel_tourism_preprocessed.csv",
    index=False
)

print("\n✅ Preprocessing completed successfully!")
print("Final Shape:", df.shape)
```
